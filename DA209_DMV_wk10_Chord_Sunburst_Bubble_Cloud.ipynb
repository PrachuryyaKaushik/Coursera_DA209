# **1. Chord Diagram**

Chord diagrams visualize flows or relationships between a set of nodes arranged in a circle.

In football, they are perfect for representing passing networks between players.

```python
import holoviews as hv
from holoviews import opts
import pandas as pd
hv.extension('bokeh')

# Sample Data: Passing network between key players in a match
passing_data = pd.DataFrame({
    'source': ['Messi', 'Messi', 'Di Maria', 'Di Maria', 'Alvarez', 'Enzo', 'Enzo'],
    'target': ['Di Maria', 'Alvarez', 'Messi', 'Alvarez', 'Messi', 'Messi', 'Di Maria'],
    'value': [15, 12, 18, 8, 5, 14, 10] # Number of successful passes
})

# Create Chord component
chord = hv.Chord(passing_data)

# Variation for Students: Customize styling and interactivity
chord.opts(
    opts.Chord(
        cmap='Set2', edge_cmap='Set2',
        edge_color=hv.dim('source').str(),
        node_color=hv.dim('index').str(),
        labels='index', width=500, height=500,
        title="Argentina Passing Network - 2022 World Cup"
    )
)
```

# **2. Sunburst Plots**

Sunburst charts display hierarchical data radiating outwards from a root node.

```python
import plotly.express as px
import pandas as pd

# Sample Data: Multi-level breakdown of goals scored
sunburst_data = pd.DataFrame({
    'Confederation': ['UEFA', 'UEFA', 'CONMEBOL', 'CONMEBOL', 'CAF', 'CAF'],
    'Team': ['Mbappé (FRA)', 'Giroud (FRA)', 'Messi (ARG)', 'Alvarez (ARG)', 'En-Nesyri (MAR)', 'Ziyech (MAR)'],
    #'Team': ['Mbappé (FRA) 8', 'Giroud (FRA) 4', 'Messi (ARG) 7', 'Alvarez (ARG) 4', 'En-Nesyri (MAR) 3', 'Ziyech (MAR) 1'],
    'Goals': [8, 4, 7, 4, 3, 1]
})

# Basic Variation
fig1 = px.sunburst(sunburst_data, path=['Confederation', 'Team'], values='Goals',
                  title="2022 World Cup Top Scorers Breakdown")
fig1.show()
```

**2b. Sunburst Plots** variation with color as per values.

```python
# Variation 2: Color mapping by continuous data to highlight impact
fig2 = px.sunburst(sunburst_data, path=['Confederation', 'Team'], values='Goals',
                  color='Goals', color_continuous_scale='RdBu',
                  title="Sunburst Variation: Color Encoded by Value")
fig2.show()
```

*Note: In Sunburst plots, coloring by values may lead to confusion. So default formatting is often better.*

# **3. Gantt Charts**
Gantt charts map activities over a timeline, often used in project management.

In case of World Cup, we can plot the progression of tournament stages.

```python
import plotly.express as px
import pandas as pd

# Sample Data: Timeline of the tournament phases
gantt_data = pd.DataFrame([
    dict(Task="Group Stage", Start='2026-06-11', End='2026-06-27', Phase="Group"),
    dict(Task="Round of 32", Start='2026-06-28', End='2026-07-03', Phase="Knockout"),
    dict(Task="Round of 16", Start='2026-07-04', End='2026-07-07', Phase="Knockout"),
    dict(Task="Quarter-Finals", Start='2026-07-09', End='2026-07-11', Phase="Knockout"),
    dict(Task="Semi-Finals", Start='2026-07-14', End='2026-07-15', Phase="Knockout"),
    dict(Task="Finals", Start='2026-07-18', End='2026-07-19', Phase="Finals")
])

# Generate Timeline
fig = px.timeline(gantt_data, x_start="Start", x_end="End", y="Task", color="Phase",
                  title="FIFA World Cup 2026 Schedule Tracking")
fig.update_yaxes(autorange="reversed") # Keeps chronological order top-to-bottom
fig.show()
```

# **4. Highlight Table**
A highlight table applies conditional color formatting to text-based data tables.

It bridges the gap between raw data analysis and visual heatmaps, helping users spot performance clusters instantly.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Data: Goals scored per match across different editions
matrix_data = {
    'Group Stage': [2.5, 2.6, 2.8],
    'Quarter-Finals': [1.8, 3.2, 2.3],
    'Semi-Finals': [3.0, 1.5, 2.0],
    'Final': [1.0, 6.0, 4.0]
}
df = pd.DataFrame(matrix_data, index=['2014', '2018', '2022'])

# Method A: Using Seaborn Heatmap for an analytical presentation
plt.figure(figsize=(8, 4))
sns.heatmap(df, annot=True, cmap="YlGnBu", cbar=True, fmt=".1f")
plt.title("Highlight Table: Average Goals per Match Phase")
plt.show()
```

# **5. Bubble Cloud**
A bubble cloud packs tightly arranged, non-overlapping circles together.

The area of each circle corresponds directly to a value.

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data: World Cup Wins per Country
teams = ['Brazil', 'Germany', 'Italy', 'Argentina', 'France', 'Uruguay']
wins = [5, 4, 4, 3, 2, 2]

# Calculate circle packaging coordinates semi-manually via random jitter distribution
x = [0, -1.2, 1.2, -0.5, 0.8, -1.5]
y = [0, 0.8, -0.7, -1.5, 1.4, -1.0]
sizes = [w * 2000 for w in wins] # Scaling factor for radius visual impact

fig, ax = plt.subplots(figsize=(8, 6))
scatter = ax.scatter(x, y, s=sizes, c=wins, cmap='viridis', alpha=0.7, edgecolors='black')

# Annotate labels into the bubbles
for i, txt in enumerate(teams):
    ax.annotate(f"{txt}\n({wins[i]})", (x[i], y[i]), ha='center', va='center', fontsize=10, weight='bold')

plt.xlim(-3, 3)
plt.ylim(-3, 3)
plt.axis('off')
plt.title("Bubble Cloud: Historical World Cup Titles")
plt.show()
```

# **5b. Bubble Cloud plot with circlify**

Using the `circlify` library is the perfect way to plot bubbles touch edges tightly without overlapping or leaving irregular gaps.

```python
import matplotlib.pyplot as plt
import circlify

# 1. Sample Data: World Cup Wins per Country
teams = ['Brazil', 'Germany', 'Italy', 'Argentina', 'France', 'Uruguay']
wins = [5, 4, 4, 3, 2, 2]

# 2. Compute the packed circle layout coordinates using circlify
# The input must be sorted or structured; circlify handles sizing natively based on data values
circles = circlify.circlify(
    wins,
    show_enclosure=False, # No outer circle will be drawn
    target_enclosure=circlify.Circle(x=0, y=0, r=1) # Packs them within a unit radius space
)

# 3. Initialize the Matplotlib Plot
fig, ax = plt.subplots(figsize=(5, 5))

# Remove axis lines and ticks for a clean canvas appearance
ax.axis('off')

# Find optimal plot boundaries based on circlify layout scaling
lim = max(
    max(
        abs(circle.x) + circle.r,
        abs(circle.y) + circle.r
    )
    for circle in circles
)
plt.xlim(-lim, lim)
plt.ylim(-lim, lim)

# 4. Draw each packed circle and map colors explicitly
# Map custom colors or match the data size dynamically using a colormap
cmap = plt.cm.viridis
min_w, max_w = min(wins), max(wins)

# circlify output matches the inverse index order of input data, so we reverse zip
for circle, team, win in zip(reversed(circles), teams, wins):
    x, y, r = circle.x, circle.y, circle.r

    # Normalize values to dynamically calculate continuous color values from colormap
    color_val = (win - min_w) / (max_w - min_w) if max_w != min_w else 0.5

    # Draw the boundary-touching circular patch
    circle_patch = plt.Circle((x, y), r, alpha=0.7, edgecolor='black', facecolor=cmap(color_val), linewidth=1.5)
    ax.add_patch(circle_patch)

    # Precisely center-align the annotations inside the calculated boundary space
    ax.annotate(
        f"{team}\n({win})",
        (x, y),
        va='center',
        ha='center',
        fontsize=10,
        weight='bold',
        color='white' if color_val > 0.6 else 'black' # Accessible contrast adjustment
    )

plt.title("Bubble Cloud: Historical World Cup Titles (Packed with Circlify)", fontsize=14, weight='bold')
plt.show()
```
